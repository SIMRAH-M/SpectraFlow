```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

# Benchmark calibration dataset
data = pd.DataFrame({
    'bagasse_g': list((0, 5, 10, 10, 15, 15, 8, 12)),
    'moringa_g': list((0, 0, 0, 2.5, 2.5, 4.0, 1.5, 3.0)),
    'flow_rate_ml_min': list((95, 72, 58, 52, 44, 38, 63, 49)),
    'removal_pct': list((24.0, 68.5, 79.2, 91.4, 94.8, 96.2, 82.1, 93.5))
})

X = data.iloc[:, 0:3]
y = data['removal_pct']

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Predict removal efficiency for an untested optimal formulation
test_recipe = pd.DataFrame({
    'bagasse_g': list((14,)),
    'moringa_g': list((3.5,)),
    'flow_rate_ml_min': list((42,))
})
pred = model.predict(test_recipe)

print(f"Model Training R² Score: {r2_score(y, model.predict(X)):.3f}")
print(f"Predicted Microplastic Removal for 14g Bagasse + 3.5g Moringa: {pred.item():.1f}%")
```
