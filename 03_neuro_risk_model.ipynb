```python
import matplotlib.pyplot as plt

def model_neuro_exposure(raw_particles_per_L=150, removal_pct=94.8, daily_intake_L=2.0):
    annual_unfiltered = raw_particles_per_L * daily_intake_L * 365
    annual_filtered = annual_unfiltered * (1 - removal_pct / 100.0)
    prevented = annual_unfiltered - annual_filtered

    print(f"Annual Unfiltered Exposure: {annual_unfiltered:,.0f} particles/year")
    print(f"Annual SpectraFlow Filtered Exposure: {annual_filtered:,.0f} particles/year")
    print(f"Total Microplastics Prevented (1 Person): {prevented:,.0f} particles/year")

    labels = ['Unfiltered Tap Water', 'SpectraFlow Filtered']
    values = [annual_unfiltered, annual_filtered]
    colors = ['#d9534f', '#2e7d32']

    plt.figure(figsize=(6, 4))
    bars = plt.bar(labels, values, color=colors, width=0.5)
    plt.ylabel('Annual Microplastic Intake (Particles / Year)')
    plt.title('Annual Neurological Microplastic Exposure Reduction')
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval + 2000, f"{yval:,.0f}", ha='center', fontweight='bold')
    plt.ylim(0, annual_unfiltered * 1.18)
    plt.tight_layout()
    plt.show()

model_neuro_exposure()
```
